# PEMBAHASAN

## 2.1 Dasar Teori
1. Python

Python adalah bahasa pemrograman tingkat tinggi yang dirancang dengan sintaks yang sederhana dan mudah dibaca. Python menggunakan pendekatan pemrograman yang memungkinkan programmer membuat program dengan kode yang relatif singkat. Python memiliki berbagai fitur dan pustaka yang dapat digunakan untuk mengembangkan berbagai jenis aplikasi. Dalam pembelajaran dasar, Python dapat digunakan untuk memahami konsep variabel, tipe data, operator, percabangan, perulangan, fungsi, dan struktur data.

2. Visual Studio Code

Visual Studio Code atau VS Code adalah editor kode sumber yang dapat digunakan untuk menulis dan mengembangkan program. VS Code mendukung berbagai bahasa pemrograman, termasuk Python. VS Code menyediakan berbagai fitur yang membantu programmer, seperti syntax highlighting, autocomplete, debugging, terminal terintegrasi, serta ekstensi. Dengan fitur tersebut, proses penulisan dan pengujian program dapat dilakukan dengan lebih mudah dan terorganisasi.

3. Git

Git adalah sistem pengontrol versi atau version control system yang digunakan untuk mencatat dan mengelola perubahan pada file dalam sebuah proyek. Git memungkinkan pengguna untuk mengetahui riwayat perubahan kode serta kembali ke versi tertentu apabila diperlukan. Dalam penggunaan Git, terdapat beberapa perintah dasar, seperti git init untuk membuat repositori lokal, git add untuk menambahkan perubahan ke staging area, git commit untuk menyimpan perubahan ke dalam riwayat Git, dan git push untuk mengirim perubahan ke repositori remote.

4. GitHub

GitHub merupakan platform berbasis web yang digunakan untuk menyimpan dan mengelola repositori Git secara daring. GitHub memungkinkan pengguna menyimpan kode program, melihat riwayat perubahan, serta melakukan kolaborasi dalam pengembangan perangkat lunak. GitHub juga dapat digunakan sebagai tempat untuk menyimpan proyek sehingga kode dapat diakses kembali dari perangkat lain selama pengguna memiliki akses ke repositori tersebut.

5. hubungan VS Code dengan GitHub

VS Code dapat digunakan bersama Git untuk mengelola proyek yang kemudian dihubungkan dengan repositori GitHub. Secara umum, proses tersebut meliputi pembuatan folder proyek, inisialisasi Git, menambahkan file ke staging area, melakukan commit, menghubungkan repositori lokal dengan repositori GitHub, dan melakukan push. Dengan alur tersebut, kode yang dibuat pada komputer dapat disimpan pada repositori GitHub dan perubahan proyek dapat dikelola secara terstruktur.

### 2.2.2 Membuat Repository GitHub

Setelah program dibuat dan diuji pada VS Code, proyek kemudian dapat dihubungkan dengan GitHub dengan menggunakan repository. Repository digunakan untuk menyimpan kode program secara daring.

Langkah-langkah:

1. Buka GitHub
2. Klik tanda + di pojok kanan atas.
3. Pilih New repository.
4. Isi informasi Repository
5. Tambahkan README untuk menjelaskan isi project.
6. Buat Repository dengan mengklik tombol: Create repository
7. memasukkan project dari VS Code

Kalau kamu sudah memiliki folder project Python di VS Code, buka terminal VS Code dan jalankan:

```python
git init
git add .
git commit -m "Initial commit"
git branch -M main
git remote add origin https://github.com/username/nama-repository.git
git push -u origin main
```

**Penjelasan :**
1. git init berfungsi untuk	Membuat repository Git baru di dalam folder project
2. git add .	berfungsi untuk Menambahkan semua file project ke staging area Git
3. git commit -m "Initial commit" berfungsi untuk Menyimpan perubahan pertama ke repository Git dengan pesan Initial commit
4. git branch -M main	berfungsi untuk Mengubah nama branch utama menjadi main
5. git remote add origin ...	berfungsi untuk Menghubungkan repository lokal dengan repository di GitHub
6. git push -u origin main	berfungsi untuk Mengunggah file dan commit dari komputer ke GitHub

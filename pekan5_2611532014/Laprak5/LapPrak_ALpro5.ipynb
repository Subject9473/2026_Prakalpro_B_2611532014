## 2.1 Dasar Teori

1. Pengertian Perulangan For

Perulangan for pada Python merupakan struktur perulangan yang digunakan untuk menjalankan suatu proses secara berulang berdasarkan sejumlah nilai atau urutan tertentu.

2. Fungsi Perulangan For

Menjalankan instruksi secara berulang berdasarkan jumlah pengulangan tertentu, menghasilkan deret angka secara berurutan, membentuk pola karakter atau angka, mengurangi penulisan kode yang sama secara berulang, membantu membuat pola dua dimensi yang terdiri dari beberapa baris dan kolom, dan juga dapat mempermudah pengolahan data yang memiliki urutan atau pola tertentu.

3. Fungsi Range pada Perulangan For

Fungsi range() digunakan untuk menghasilkan urutan bilangan yang dapat digunakan sebagai sumber pengulangan pada for. Salah satu bentuk penggunaannya adalah menentukan nilai awal dan batas akhir. Misalnya, perulangan dari 1 sampai batas tertentu dapat digunakan untuk menentukan jumlah baris yang akan dibuat. Perulangan luar menggunakan pola range(1, batas + 1) sehingga perulangan berjalan dari baris pertama sampai baris terakhir.

4. Perulangan untuk Membentuk Pola

Perulangan for dapat digunakan untuk membentuk pola yang terdiri dari angka atau karakter. Setiap pengulangan dapat menghasilkan bagian tertentu dari pola.

5. Outer Loop dan Inner Loop

Dalam pembuatan pola yang kompleks, perulangan dapat dibagi menjadi outer loop dan inner loop. Outer loop atau perulangan luar berfungsi sebagai perulangan vertikal untuk setiap baris. Sebagai contoh, jika sebuah pola terdiri dari lima baris, maka outer loop akan berjalan sebanyak lima kali. Setiap satu kali pengulangan outer loop mewakili satu baris pola. Sedangkan, Inner loop atau perulangan dalam digunakan sebagai perulangan horizontal untuk membentuk pola di dalam setiap baris. Inner loop dapat digunakan untuk mencetak karakter yang jumlahnya berbeda pada setiap baris.

6. Nested For Loop

Nested for loop adalah perulangan for yang berada di dalam perulangan for lainnya. Struktur ini digunakan ketika suatu pola memiliki dua dimensi, yaitu baris dan isi dari setiap baris.

7. Penggunaan Parameter End

Dalam pembuatan pola menggunakan perulangan end="" digunakan agar karakter berikutnya dicetak pada baris yang sama. Hal ini penting ketika beberapa karakter harus disusun secara horizontal. Setelah seluruh karakter dalam satu baris selesai dicetak, print() tanpa argumen dapat digunakan untuk berpindah ke baris berikutnya.

8. Pemetaan Perulangan dengan Angka

Perulangan for juga dapat digunakan untuk menghasilkan deret angka berdasarkan suatu rumus. Pada praktikum menunjukkan bahwa pola angka dapat dianalisis dengan melihat hubungan antara nilai penghitung dan angka yang ingin dicetak. Contohnya adalah pola:

4 7 10 13 16

yang memiliki kenaikan sebesar 3 pada setiap langkah. Hubungan antara nomor pengulangan dan angka yang dihasilkan dapat digunakan untuk menentukan rumus perhitungan angka.

9. Loop Table

Loop table atau tabel perulangan merupakan metode untuk menganalisis hubungan antara nilai penghitung dan nilai yang dihasilkan oleh program.

Sebagai contoh, untuk menghasilkan:

2 7 12 17 22

dapat dibuat tabel yang menunjukkan nilai count dan angka yang dihasilkan. Setiap kenaikan count sebesar 1 menyebabkan angka bertambah sebesar 5. Dari pola tersebut dapat ditentukan hubungan matematis antara count dan angka yang dicetak. Loop table membantu programmer memahami pola sebelum membuat kode program.

10. Pola Angka Menurun

Perulangan for tidak hanya dapat menghasilkan angka yang meningkat, tetapi juga dapat menghasilkan angka yang menurun. Contoh pola:

17 13 9 5 1

Pada pola tersebut, setiap angka berkurang sebesar 4 dari angka sebelumnya. Dengan membuat tabel perulangan, hubungan antara nilai penghitung dan angka yang dicetak dapat diketahui sehingga rumus yang sesuai dapat dibuat.

11. Perulangan untuk Membentuk Pola Dua Dimensi

Pola dua dimensi terdiri dari beberapa baris dan setiap baris dapat memiliki susunan karakter yang berbeda. Oleh karena itu, pola seperti ini membutuhkan pengaturan antara baris dan isi setiap baris. Outer loop digunakan untuk menentukan baris, sedangkan inner loop digunakan untuk menentukan jumlah karakter yang ditampilkan pada setiap baris. Konsep ini dapat digunakan untuk membuat berbagai bentuk pola seperti segitiga, piramida, persegi, pola angka, hingga pola geometris yang lebih kompleks.

12. Hubungan Nomor Baris dengan Jumlah Karakter

Dalam pembuatan pola, jumlah karakter pada setiap baris sering kali memiliki hubungan matematis dengan nomor baris. Contoh pola:

....1

...2

..3

.4

5

jumlah titik pada setiap baris adalah:

* Baris 1 = 4 titik
* Baris 2 = 3 titik
* Baris 3 = 2 titik
* Baris 4 = 1 titik
* Baris 5 = 0 titik

Dari pola tersebut dapat diketahui bahwa jumlah titik berkurang satu setiap berpindah ke baris berikutnya melalui perhitungan -1 * line + 5.

13. Konsep Nested Loop dalam Pembuatan Pola Kompleks

Nested loop sangat berguna ketika setiap baris memiliki pola yang berbeda.Pada praktikum, outer loop berjalan dari baris pertama sampai baris terakhir. Kemudian inner loop menentukan jumlah titik yang harus dicetak berdasarkan nomor baris. Setelah titik selesai dicetak, angka baris ditampilkan. Konsep tersebut dapat dikembangkan untuk membuat pola yang lebih kompleks, seperti pola angka simetris, pola piramida, pola jam pasir, dan pola lainnya.

### 2.2.1 Program "jumlah_genap_2611532014"


```python
#program perulangan for

ulang_2014 = int(input("masukkan nilai batas: "))

jumlah_2014 = 0
for i_2014 in range(1, ulang_2014 + 1):
    if i_2014 % 2 == 0:
        print(i_2014, end=" ")
        jumlah_2014 = jumlah_2014 + i_2014

        if i_2014 < ulang_2014:
            print(" + ", end="")
        else:
            print(" = ", jumlah_2014, end="")
print()
print("jumlah =", jumlah_2014)
```

**Penjelasan :**
1. Meminta pengguna memasukkan nilai batas perulangan. input() menerima masukan dalam bentuk teks. int() mengubah masukan menjadi bilangan bulat. Nilai disimpan dalam variabel ulang_2014.
2. Variabel jumlah_2014 digunakan untuk menyimpan hasil penjumlahan bilangan genap dengan nilai awal 0.
3. for digunakan untuk melakukan perulangan. range(1, ulang_2014 + 1) menghasilkan bilangan mulai dari 1 sampai nilai ulang_2014. Variabel i_2014 akan berubah mengikuti setiap angka dalam rentang tersebut.
4. Operator % digunakan untuk mencari sisa hasil pembagian. Jika i_2014 % 2 == 0, berarti bilangan tersebut habis dibagi 2 sehingga merupakan bilangan genap.
5. end=" " membuat hasil berikutnya tetap berada pada baris yang sama dan dipisahkan spasi.
6. Setiap bilangan genap ditambahkan ke jumlah_2014 sehingga variabel tersebut terus menyimpan total penjumlahan bilangan genap.
7. print(" + ", end="") dan print(" = ", jumlah_2014, end="") digunakan untuk membuat tampilan hasil penjumlahan menjadi: 2 + 4 + 6 = 12. Jika i_2014 masih lebih kecil dari batas, program mencetak tanda +. Jika sudah mencapai kondisi else, program mencetak tanda = diikuti hasil jumlah.
8. print() Digunakan untuk membuat baris baru setelah proses perulangan selesai.
9. print("jumlah =", jumlah_2014) Menampilkan total seluruh bilangan genap yang telah dihitung.

### 2.2.3 Program "nested_for1_2611532014"

```python
#program perulangan for

batas_2014 = int(input("masukkan nilai batas: "))
for line in range(1, batas_2014 + 1):
    for j_2014 in range(1, (-1 * line + batas_2014) + 1):
        print(" . ", end=" ")
    print(line)
```

**Penjelasan :**
1. Meminta pengguna memasukkan nilai batas. input() menerima masukan dari pengguna.
int() mengubah masukan menjadi bilangan bulat. Nilai disimpan dalam variabel batas_2014.
2. for line in range(1, batas_2014 + 1): digunakan untuk menentukan baris yang akan ditampilkan. line dimulai dari 1 sampai batas_2014. Jika batas yang dimasukkan adalah 5, maka nilai line adalah 1, 2, 3, 4, 5.
3. for j_2014 in range(1, (-1 * line + batas_2014) + 1): Perulangan ini menentukan berapa banyak titik yang dicetak pada setiap baris. Rumus (-1 * line + batas_2014) dapat ditulis sebagai: batas_2014 - line. Karena nilai line semakin bertambah, jumlah titik semakin berkurang.
4. print(" . ", end=" ") Mencetak karakter titik pada baris yang sedang diproses. end=" " membuat titik berikutnya dicetak pada baris yang sama.
5. print(line) Setelah semua titik pada suatu baris selesai dicetak, program mencetak nilai line. print() juga membuat program berpindah ke baris berikutnya.

### 2.2.6 Program "nested_for4_2611532014"

```python
#program perulangan for

tinggi_2014 = int(input("masukkan tinggi pola (bilangan genap, misal 10): "))


if tinggi_2014 % 2 != 0:
    print("tinggi harus bilangan genap!")
else:
    a_2014 = tinggi_2014
    c_2014 = a_2014
    lebar_2014 = (2 * tinggi_2014) - 2

    for i_2014 in range(1, tinggi_2014 + 1):
        b_2014 = c_2014 + 1

        for j_2014 in range(1, lebar_2014 + 1):

            #baris atas dan bawah
            if i_2014 == 1 or i_2014 == tinggi_2014:
                 if j_2014 == 1 or j_2014 == lebar_2014:
                     print("#", end="")
                 else:
                     print("=", end="")
            #baris isi
            else:
                if j_2014 == 1 or j_2014 == lebar_2014:
                    print("|", end="")
                else: 
                    if j_2014 == c_2014:
                        print("<", end="")
                    elif j_2014 == b_2014:
                        print(">", end="")
                    elif j_2014 == (lebar_2014 - c_2014):
                        print("<", end="")
                    elif j_2014 == (lebar_2014 - c_2014 + 1):
                        print(">", end="")
                    elif j_2014 > b_2014 and j_2014 < (lebar_2014 - c_2014):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()
        #logika asli java
        a_2014 -= 2

        if a_2014 <= 0:
            c_2014 = (-a_2014) + 2
        else:
            c_2014 = a_2014
```

5. b_2014 = c_2014 + 1 b_2014 digunakan sebagai posisi yang berdekatan dengan c_2014. Kedua nilai ini nanti akan digunakan untuk menentukan posisi karakter < dan > di bagian dalam pola.
6. for j_2014 in range(1, lebar_2014 + 1): merupakan inner loop. j_2014 menunjukkan posisi karakter dari kiri ke kanan. Jika lebar_2014 = 18, maka setiap baris diproses sebanyak 18 posisi.    * if j_2014 == 1 or j_2014 == lebar_2014: dan print("#", end="") Jika posisi berada di paling kiri atau paling kanan, program mencetak: # sebagai bagian sudut/batas pola.
   * else: print("=", end="") Jika bukan posisi paling kiri atau kanan, program mencetak =.
8. Jika bukan baris pertama atau terakhir program masuk ke bagian isi pola. if j_2014 == 1 or j_2014 == lebar_2014: dan print("|", end="") Posisi paling kiri dan kanan dicetak dengan | yang berfungsi sebagai batas samping pola.
9. Menentukan posisi karakter <. if j_2014 == c_2014: print("<", end="") Jika posisi j_2014 sama dengan c_2014, program mencetak <. Lalu, elif j_2014 == (lebar_2014 - c_2014): dan print("<", end="") menentukan posisi < lainnya berdasarkan posisi simetris terhadap lebar pola.
10. Menentukan posisi karakter >. elif j_2014 == b_2014: dan print(">", end="") Jika posisi sama dengan b_2014, program mencetak >. Lalu elif j_2014 == (lebar_2014 - c_2014 + 1): dan print(">", end="") menghasilkan posisi > yang berpasangan secara simetris.
11. elif j_2014 > b_2014 and j_2014 < (lebar_2014 - c_2014): dan print(".", end="") Jika posisi berada di antara batas b_2014 dan posisi simetrisnya, program mencetak titik . yang digunakan untuk membentuk area tengah pola.
12. else: print(" ", end="") Mencetak spasi, jika tidak memenuhi kondisi sebelumnya

### 2.2.7 Program "perulangan_for1_2611532014"

### 2.2.9 Program "perulangan_for3_2611532014"

```python
#program perulangan for

ulang_2014 = int(input("masukkan jumlah perulangan: "))

jumlah_2014 = 0
for i_2014 in range(1, ulang_2014+1):
    print(i_2014, end=" ")
    jumlah_2014 = jumlah_2014 + i_2014

    if i_2014 < ulang_2014:
        print("+", end=" ")
    else:
        print("=", jumlah_2014, end=" ")
print()
print("jumlah =", jumlah_2014)
```

**Penjelasan :**
1. ulang_2014 = int(input("masukkan jumlah perulangan: ")) Program meminta pengguna memasukkan jumlah perulangan.
2. jumlah_2014 = 0 digunakan untuk menyimpan hasil penjumlahan dengan nilai awal 0.
3. for i_2014 in range(1, ulang_2014+1): Perulangan dimulai dari angka 1. Batas akhirnya menggunakan ulang_2014 + 1 agar nilai ulang_2014 ikut diproses.
4. print(i_2014, end=" ") Menampilkan angka yang sedang diproses. end=" " membuat angka berikutnya tetap berada pada baris yang sama.
5. jumlah_2014 = jumlah_2014 + i_2014 Setiap nilai i_2014 ditambahkan ke jumlah_2014.
6. if i_2014 < ulang_2014: print("+", end=" ") dan else: print("=", jumlah_2014, end=" ") Jika angka yang sedang diproses belum mencapai angka terakhir, program mencetak tanda +. Jika sudah mencapai angka terakhir, program mencetak tanda = dan hasil penjumlahan.
7. print("jumlah =", jumlah_2014) Menampilkan hasil penjumlahan akhir yang tersimpan dalam jumlah_2014.

# KESIMPULAN

Berdasarkan praktikum yang telah dilakukan, dapat disimpulkan bahwa perulangan for pada Python merupakan struktur yang digunakan untuk menjalankan perintah secara berulang berdasarkan jumlah atau rentang tertentu. Perulangan for dapat digunakan untuk menghasilkan pola angka maupun karakter dengan memanfaatkan fungsi range(). Dalam pembuatan pola yang lebih kompleks, perulangan for dapat dikombinasikan menjadi nested for loop, yaitu perulangan di dalam perulangan. Outer loop digunakan untuk mengatur jumlah baris atau bagian utama pola, sedangkan inner loop digunakan untuk mengatur karakter atau elemen yang terdapat pada setiap baris. Selain itu, penggunaan loop table membantu dalam menentukan hubungan antara nilai penghitung (count) dengan angka yang akan ditampilkan. Dengan mengamati perubahan nilai pada setiap perulangan, pola matematika dapat ditemukan sehingga program dapat menghasilkan urutan angka yang sesuai.

Dengan demikian, praktikum perulangan for memberikan pemahaman mengenai cara menggunakan perulangan sederhana maupun nested for loop untuk membuat berbagai pola dan urutan. Pemahaman terhadap hubungan antara jumlah perulangan, nilai range(), serta pola pada setiap baris menjadi dasar penting dalam menyusun program Python yang dapat menghasilkan keluaran secara terstruktur dan sesuai dengan kebutuhan.

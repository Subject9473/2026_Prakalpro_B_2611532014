## 2.1 Dasar Teori

Struktur percabangan merupakan salah satu konsep dasar dalam pemrograman yang digunakan untuk mengatur jalannya program berdasarkan kondisi tertentu. Dengan struktur percabangan, program dapat menentukan perintah yang harus dijalankan apabila suatu kondisi terpenuhi atau tidak terpenuhi. 

1. Percabangan if

Percabangan if merupakan bentuk paling sederhana dari struktur percabangan pada Python. Pernyataan ini digunakan untuk menjalankan suatu kode hanya apabila kondisi yang diberikan bernilai benar (True). Jika kondisi bernilai salah (False), kode tersebut tidak akan dijalankan dan program akan melanjutkan ke instruksi berikutnya.

2. Percabangan if-else

Percabangan if-else digunakan ketika terdapat dua kemungkinan tindakan. Jika kondisi bernilai benar, program menjalankan blok if. Sebaliknya, apabila kondisi bernilai salah, program menjalankan blok else.

3. Percabangan if-elif-else

Percabangan if-elif-else digunakan apabila program memiliki lebih dari dua kemungkinan kondisi. Python akan memeriksa kondisi mulai dari if, kemudian berlanjut ke elif apabila kondisi sebelumnya tidak terpenuhi. Jika seluruh kondisi if dan elif tidak terpenuhi, blok else akan dijalankan.

4. Multi-if

Multi-if merupakan penggunaan beberapa pernyataan if secara terpisah. Berbeda dengan if-elif-else, setiap kondisi pada multi-if diperiksa secara independen. Oleh karena itu, lebih dari satu kondisi dapat terpenuhi dan masing-masing blok perintah dapat dijalankan.

5. Match-case

Match-case merupakan struktur pemilihan yang tersedia pada Python untuk mencocokkan suatu nilai dengan beberapa pola atau kemungkinan. Struktur ini diperkenalkan pada Python 3.10 dan dapat digunakan sebagai alternatif ketika program memiliki banyak pilihan berdasarkan suatu nilai.

6. Nested if

Nested if adalah struktur percabangan yang menempatkan pernyataan if di dalam blok percabangan lainnya. Struktur ini digunakan ketika suatu kondisi kedua hanya perlu diperiksa setelah kondisi pertama terpenuhi.

7. Operator Logika dalam Percabangan

Struktur percabangan dapat dikombinasikan dengan operator logika untuk membuat kondisi yang lebih kompleks. Beberapa operator logika yang umum digunakan dalam Python adalah:

* and, menghasilkan True apabila seluruh kondisi yang digabungkan bernilai benar.
* or, menghasilkan True apabila setidaknya salah satu kondisi bernilai benar.
* not, membalik nilai logika suatu kondisi.

## 2.2 Kode Program

### 2.2.1 Program "if_elif_else1_2611532014"


```python
#program kondisional if

umur_2014 = int(input("Input Umur Anda: "))
sim_2014 = input("Apakah Anda Sudah Punya Sim C (y/t): ")[0]

if umur_2014 >= 17 and sim_2014 == 'y':
    print("Anda Sudah dewasa dan boleh bawa motor")
elif umur_2014 >= 17 and sim_2014 != 'y':
    print("Anda Sudah dewasa tetapi tidak boleh bawa motor") 
elif umur_2014 < 17 and sim_2014 == 'y':
    print("Anda belum cukup umur punya sim")
else:
    print("Anda belum cukup umur dan tidak boleh bawa motor")
print("Program Selesai")
```

**Penjelasan :**
1. Fungsi input() menghasilkan data bertipe string, sehingga digunakan int() untuk mengubahnya menjadi bilangan bulat.
2. Program meminta pengguna memasukkan y jika memiliki SIM C atau t jika tidak memiliki SIM C. Bagian [0] digunakan untuk mengambil karakter pertama dari input.
3. Program ini menggunakan operator logika and.
    * Pada kondisi pertama berlaku apabila user berumur minimal 17 tahun dan memiliki SIM C (y). Sehingga program akan menampilkan bahwa pengguna sudah dewasa dan boleh membawa motor.
    * Kondisi kedua berlaku apabila user berumur sudah 17 tahun atau lebih dan tidak memiliki SIM C (t). Operator != berarti tidak sama dengan. Jadi, pengguna dianggap sudah cukup umur tetapi belum memiliki SIM C.
    * Kondisi ini berlaku apabila user berumur kurang dari 17 tahun dan pengguna memasukkan y untuk kepemilikan SIM. Sehingga, program memberikan informasi bahwa pengguna belum cukup umur untuk memiliki SIM.
    * else dijalankan apabila seluruh kondisi sebelumnya bernilai salah. Dalam kondisi ini, pengguna berumur kurang dari 17 tahun dan tidak memiliki SIM C.

### 2.2.2 Program "if1_2611532014"

### 2.2.4 Program "match_case1_2611532014"

```python
#rogram match case
#konversi nilai angka ke nama bulan

bulan_2014 = int(input("masukkan angka bulan (1 - 12): "))

match bulan_2014:
    case 1:
        print("Januari")
    case 2:
            print("Februari")
    case 3:
            print("Maret")
    case 4:
            print("April")
    case 5:
            print("May")
    case 6:
            print("Juni")
    case 7:
            print("Juli")
    case 8:
            print("Agustus")
    case 9:
            print("September")
    case 10:
            print("Oktober")
    case 11:
            print("November")
    case 12:
            print("Desember")
    case _:
            print("Angka Tidak Valid")
```

**Penjelasan :**
1. input() digunakan untuk menerima masukan.
2. int() mengubah masukan menjadi bilangan bulat.
3. match digunakan untuk membandingkan nilai yang terdapat pada variabel bulan_2014 dengan beberapa pilihan yang tersedia pada case.
4. Setiap case mewakili satu bulan. contoh: case 8 = Agustus
5. Tanda _ (case _) berfungsi sebagai default case. Bagian ini akan dijalankan apabila nilai yang dimasukkan tidak cocok dengan case 1 sampai 12.

### 2.2.5 Program "multi_if1_2611532014"

```python
#program untuk kondisional if

umur_2014 = int(input("Input Umur Anda: "))
sim_2014 = input("Apakah Anda Sudah Punya Sim C (y/t): ")[0]

if umur_2014 >= 17 and sim_2014 == 'y':
    print("Anda Sudah dewasa dan boleh bawa motor")

if umur_2014 >= 17 and sim_2014 != 'y':
    print("Anda Sudah dewasa tetapi tidak boleh bawa motor")
  
if umur_2014 < 17 and sim_2014 == 'y':
    print("Anda belum cukup umur punya sim")

if umur_2014 < 17 and sim_2014 != 'y':
    print("Anda belum cukup umur bawa motor")
```

**Penjelasan :**
1. input() menerima data dari pengguna.
2. int() mengubah input menjadi bilangan bulat.
3. [0] digunakan untuk mengambil karakter pertama dari input.
4. Operator and berarti kedua kondisi harus benar. Kondisi pertama terpenuhi apabila user berumur minimal 17 tahun, dan memiliki SIM C.
5. Operator != berarti tidak sama dengan. Kondisi kedua terpenuhi apabila user berumur minimal 17 tahun, dan SIM C bukan y.
6. Kondisi ketiga terpenuhi apabila user berumur kurang dari 17 tahun, dan pengguna memasukkan y untuk SIM C.
7. Kondisi keempat terpenuhi apabila user berumur kurang dari 17 tahun, dan pengguna tidak memiliki SIM C.
8. Setiap kondisi diperiksa secara independen. Pada kasus ini, karena keempat kondisi dibuat saling berlawanan, hanya satu kondisi yang akan terpenuhi untuk kombinasi umur dan SIM tertentu.

### 2.2.6 Program "multi_if2_2611532014"

```python
#multi if
#input dari user
total_belanja_2014 = float(input("masukkan total belanja (Rp): "))

#input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_2014 = input("apakah anda member? (y/t): ").strip().lower()
is_member_2014 = input_member_2014 in ["y", "ya"]

#inut status kode romo (mengecek aakah user mengetik 'ya' atau 'y')
input_promo_2014 = input("apakah kode promo valid (y/t): ").strip().lower()
kode_promo_valid_2014 = input_promo_2014 in ["y", "ya"]

total_diskon_persen_2014 = 0

#mult if terpisah: setia kondisi dieriksa secara indeenden
#diskon bisa di tumuk jika ya gitulah

if total_belanja_2014 > 1000000:
    total_diskon_persen_2014 += 10 #diskon belanja besar

if is_member_2014:
    total_diskon_persen_2014 += 5 #diskon member

if kode_promo_valid_2014:
    total_diskon_persen_2014 += 15 #diskon voucher

#menghitung nominal diskon dan total bayar
nominal_diskon_2014 = total_belanja_2014 * (total_diskon_persen_2014 / 100)
total_bayar_2014 = total_belanja_2014 - nominal_diskon_2014

# output hasil
print("\n--- Rincian pembayaran ---")
print(f"total diskon : {total_diskon_persen_2014}% (Rp {nominal_diskon_2014:,.0f})")
print(f"total bayar : Rp {total_bayar_2014:,.0f})")

print(f"total diskon yang anda dapatkan: {total_diskon_persen_2014}%")
#output: total diskon yang anda daatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

**Penjelasan :**
1. input() digunakan untuk menerima masukan.
2. float() digunakan agar nilai dapat berupa angka desimal.
3. strip() digunakan untuk menghapus spasi yang tidak diperlukan, sedangkan lower() mengubah input menjadi huruf kecil.
4. is_member_2014 = input_member_2014 in ["y", "ya"] digunakan untuk memeriksa apakah pengguna memasukkan y atau ya.
    * Jika memasukkan y atau ya → is_member_2014 bernilai True.
    * Jika memasukkan t → is_member_2014 bernilai False.
5. total_diskon_persen_2014 = 0 Variabel ini digunakan untuk menyimpan total persentase diskon yang diperoleh pengguna. Nilai awalnya adalah 0%.
6. Jika total belanja lebih dari Rp1.000.000, pengguna mendapatkan diskon 10%.
7. Jika pengguna merupakan member, maka mendapatkan tambahan diskon 5%. jika sebelumnya sudah mendapatkan 10%, maka total diskon: 10% + 5% = 15%
8. Jika kode promo valid, pengguna mendapatkan tambahan diskon 15%. jika sebelumnya sudah mendapatkan 15%, maka total diskon: 15% + 15% = 30%.  diskon akumulatif yaitu Setiap if berdiri sendiri sehingga diskon dapat ditambahkan apabila beberapa kondisi terpenuhi 
9. nominal_diskon_2014 = total_belanja_2014 * (total_diskon_persen_2014 / 100) Baris tersebut menghitung jumlah uang yang didapatkan sebagai diskon.
10. total_bayar_2014 = total_belanja_2014 - nominal_diskon_2014 Total pembayaran diperoleh dengan mengurangi total belanja dengan nominal diskon.
11. :,.0f digunakan untuk menampilkan angka tanpa angka desimal dan menggunakan pemisah ribuan.

### 2.2.7 Program "nilai_2611532014"

```python
#rogram nested if
#konversi nilai angka ke huruf

nilai_2014 = int(input("inputkan nilai angka = "))

if nilai_2014 >= 81:
    print("A")
elif nilai_2014 >= 70:
    print("B")
elif nilai_2014 >= 60:
    print("C")
elif nilai_2014 >= 50:
    print("D")
else:
    print("E")
```

3. Kondisi nilai:

    * Jika nilai berada pada 60–69, program menampilkan: C
    * Jika nilai berada pada 50–59, program menampilkan: D
    * Jika seluruh kondisi sebelumnya tidak terpenuhi, berarti nilai di bawah 50, sehingga program menampilkan: E

# KESIMPULAN

Berdasarkan praktikum mengenai Struktur Percabangan pada Python, dapat disimpulkan bahwa struktur percabangan merupakan salah satu bagian penting dalam pemrograman yang berfungsi untuk mengatur jalannya program berdasarkan kondisi tertentu. Melalui penggunaan if, if-else, dan if-elif-else, program dapat menentukan tindakan yang berbeda sesuai dengan kondisi yang terpenuhi. Sementara itu, multi-if dapat digunakan untuk memeriksa beberapa kondisi secara terpisah sehingga beberapa perintah dapat dijalankan apabila kondisinya terpenuhi. Selain itu, match-case dapat digunakan untuk melakukan pemilihan berdasarkan nilai atau pola tertentu. Nested if juga dapat digunakan untuk melakukan pemeriksaan kondisi secara bertingkat dengan menempatkan percabangan di dalam percabangan lainnya. Dengan memahami berbagai struktur percabangan tersebut, pengguna dapat menyusun logika program Python yang lebih terarah, sistematis, dan sesuai dengan kondisi yang diberikan.

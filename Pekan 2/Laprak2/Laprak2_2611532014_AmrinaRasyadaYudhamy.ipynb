## 2.1 Dasar Teori

1. Boolean
   Boolean merupakan tipe data yang digunakan untuk menyimpan nilai logika dalam pemrograman. Tipe data Boolean hanya memiliki dua nilai, yaitu `True` yang berarti benar dan `False` yang berarti salah. Boolean biasanya digunakan dalam proses pengambilan keputusan, seperti percabangan dan perulangan. Contohnya, `status = True` menunjukkan bahwa variabel `status` memiliki nilai benar.

2. Input
   Input merupakan proses untuk menerima data atau informasi dari pengguna saat program sedang dijalankan. Dalam Python, fungsi `input()` digunakan untuk meminta pengguna memasukkan suatu data. Data yang diperoleh dari `input()` secara default bertipe string. Contohnya, `nama = input("Masukkan nama: ")` digunakan untuk menerima nama yang dimasukkan oleh pengguna.

3. Konstanta
   Konstanta adalah nilai yang digunakan dalam program dan pada dasarnya tidak dimaksudkan untuk diubah selama program berjalan. Python tidak memiliki kata kunci khusus untuk membuat konstanta, sehingga biasanya konstanta ditulis menggunakan huruf kapital sebagai tanda bahwa nilainya sebaiknya tidak diubah. Contohnya, `PI = 3.14` dapat digunakan untuk menyimpan nilai tetap yang diperlukan dalam perhitungan.

4. Numerik
   Tipe data numerik digunakan untuk menyimpan data berupa angka yang dapat digunakan dalam berbagai operasi matematika. Dalam Python, tipe data numerik yang umum digunakan adalah `int` untuk bilangan bulat dan `float` untuk bilangan desimal. Contohnya, `umur = 17` menggunakan tipe `int`, sedangkan `tinggi = 165.5` menggunakan tipe `float`. Data numerik dapat digunakan untuk operasi seperti penjumlahan, pengurangan, perkalian, dan pembagian.

5. String
   String merupakan tipe data yang digunakan untuk menyimpan teks atau kumpulan karakter. String dalam Python ditulis menggunakan tanda kutip tunggal (`'...'`) atau tanda kutip ganda (`"..."`). Contohnya, `nama = "Amrina"` merupakan variabel bertipe string. String dapat digunakan untuk menyimpan berbagai informasi berupa teks, seperti nama, alamat, kalimat, dan pesan yang akan ditampilkan oleh program.


## 2.2 Kode Program

### 2.2.1 Program "Boolean_2611532014"


```python
#buat file dengan nama boolean_nim.py
#nama variabel ditambah 4 digit nim terakhir contoh: nilai_2014
#deklarasi variable dengan tipe data boolean
is_lulus = True
is_cumlaude = True

#menggunakan boolean
nilai_2014 = 85
batas_lulus = 75

#menentukan nilai bolean dari kondisi
status_kelulusan = nilai_2014 >= batas_lulus #hasilnya akan true

print("=== check kelulusan ===")
print("Nilai:", nilai_2014)
print("apakah lulus?:", status_kelulusan)
if is_lulus and is_cumlaude:
    print("Selamat, anda lulus dengan predikat Cum Laude!")
```

**Penjelasan :**
1. is_lulus = True menunjukkan bahwa status lulus bernilai benar dan is_cumlaude = True menunjukkan bahwa status Cum Laude bernilai benar.
2. nilai_2014 menyimpan nilai yang diperoleh, yaitu 85 dan batas_lulus menyimpan nilai minimum untuk dinyatakan lulus, yaitu 75. Karena nilai 85 lebih besar atau sama dengan 75, maka siswa memenuhi batas kelulusan.
3. status_kelulusan = nilai_2014 >= batas_lulus. Baris ini membandingkan nilai dengan batas kelulusan. Maka variabel status_kelulusan memiliki nilai True.
4. Perintah print() digunakan untuk menampilkan informasi ke layar.
5. Bagian if digunakan untuk membuat kondisi. Operator and berarti kedua kondisi harus bernilai True agar perintah di dalam if dijalankan.

### 2.2.2 Program "input_2611532014"



```python
#buat file dengan nama input_nim.py
#buat program ntuk menghitung luas persegi panjang
#nama variabel ditambah 4 digit nim terakhir contoh: panjang_2014
#program ini menggunakan fungsi input() untuk meminta pengguna
#memasukkan nilai panjang dan lebar persegi panjang.
#nilai yang dimasukkan akan dikonversi menjadi tipe data integer
panjang_2014 = input('masukkan nilai panjang:')
lebar_2014 = input('masukkan nilai lebar:')
luas_2014 = int(panjang_2014) * int(lebar_2014)
print("luas =", luas_2014)
```

**Penjelasan :**
1. Fungsi input() digunakan untuk meminta pengguna memasukkan nilai panjang persegi panjang. Nilai yang dimasukkan oleh pengguna melalui input() secara default dianggap sebagai string (teks)
2. Fungsi int() digunakan untuk mengubah data yang awalnya berupa string menjadi bilangan bulat (integer) agar dapat digunakan dalam operasi perkalian.
3. Perintah print() digunakan untuk menampilkan hasil perhitungan luas ke layar.

### 2.2.3 Program "Konstanta_2611532014"

```python
#buat file dengan nama konstanta_nim.py
#program ini menggunakan konstanta untuk menghitung luas lingkaran
#nama variabel ditambah 4 digit nim terakhir contoh: jari_2014


from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_2014 = float(input('masukkan nilai jari-jari: '))
luas_2014 = PI * jari_2014 * jari_2014
print("luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_2014, luas_2014))
```

### 2.2.4 Program "numerik_2611532014"

```python
# buat file dengan nama Numerik_NIM.py
# nama variabel ditambah 4 digit terakhir NIM contoh: a_2014
# create a variable with integer value.
a_2014=100
print("the type of variable having value", a_2014, "is", type(a_2014))

#create a variable with float value.
b_2014=10.2345
print("the type of variable having value", b_2014, "is", type(b_2014))

#create a variable with complex value.
c_2014=100+3j
print("the type of variable having value", c_2014, "is", type(c_2014))
```

## 2.1 Dasar Teori

1. Konsep Dasar Operator
Dalam bahasa pemrograman Python, operator adalah simbol atau karakter khusus yang digunakan untuk melakukan operasi tertentu terhadap satu atau beberapa nilai yang disebut operan. Berdasarkan jumlah operannya, operator dapat dibedakan menjadi dua, yaitu: Operator Unary (Membutuhkan satu operan) dan Operator Binary (Membutuhkan dua operan). Setiap operator di Python memiliki tingkat prioritas (operator precedence) yang menentukan urutan eksekusi ketika beberapa operator digunakan secara bersamaan.

2. Jenis-Jenis Operator pada Python
   a. Operator Aritmatika digunakan untuk melakukan operasi matematika dasar pada tipe data numerik (int, float). Seperti Penjumlahan (+), Pengurangan (-), Perkalian (*), Pembagian (/), Pembagian Bulat (//), Modulus (%), dan Perpangkatan (**)
   
   b. Operator perbandingan digunakan untuk membandingkan dua nilai dan selalu menghasilkan nilai boolean 
   * Sama dengan (==): Bernilai True jika kedua operan bernilai sama.
   * Tidak sama dengan (!=): Bernilai True jika kedua operan tidak bernilai sama.
   * Lebih besar dari (>): Bernilai True jika operan kiri lebih besar dari operan kanan.
   * Lebih kecil dari (<): Bernilai True jika operan kiri lebih kecil dari operan kanan.
   * Lebih besar atau sama dengan (>=): Bernilai True jika operan kiri lebih besar dari atau sama dengan operan kanan.
   * Lebih kecil atau sama dengan (<=): Bernilai True jika operan kiri lebih kecil dari atau sama dengan operan kanan.
   
   c. Operator logika digunakan untuk mengombinasikan satu atau lebih ekspresi boolean.
   * and: Bernilai True hanya jika kedua kondisi/operan bernilai True.
   * or: Bernilai True jika salah satu kondisi/operan bernilai `True`.
   * not: Membalikkan nilai boolean operan (mengubah True menjadi False dan sebaliknya).
   
   d. Operator Augmented Assignment menggabungkan operasi aritmatika atau bitwise ke variabel secara singkat.
   * += (contoh: x += 5 setara dengan x = x + 5)
   * -= (contoh: x -= 3 setara dengan x = x - 3)
   * *= (contoh: x *= 2 setara dengan x = x * 2)
   * /= (contoh: x /= 4 setara dengan x = x / 4)
   * //= (contoh: x //= 2 setara dengan x = x // 2)
   * %= (contoh: x %= 3 setara dengan x = x % 3)
   * **= (contoh: x **= 2 setara dengan x = x ** 2)
   
   e. Operator keanggotaan digunakan untuk menguji apakah suatu nilai terdapat di dalam suatu struktur data koleksi (sequence) seperti string, list, tuple, set, atau dictionary.
   * in: Bernilai True jika elemen ditemukan di dalam koleksi.
   * not in: Bernilai True jika elemen tidak ditemukan di dalam koleksi.
   
   f. Operator identitas digunakan untuk membandingkan lokasi memori dari dua objek.
   * is: Bernilai True jika kedua variabel merujuk pada objek yang sama di memori.
   * is not: Bernilai True jika kedua variabel merujuk pada objek yang berbeda di memori.
   
   g. Operator bitwise bekerja secara langsung pada level biner dari bilangan bulat. Nilai desimal diubah menjadi format biner terlebih dahulu sebelum operasi dijalankan.
   * AND (&): Membandingkan setiap bit; bernilai 1 jika kedua bit bernilai 1.
   * OR (|): Membandingkan setiap bit; bernilai 1 jika salah satu atau kedua bit bernilai 1.
   * XOR (^): Membandingkan setiap bit; bernilai 1 jika hanya salah satu bit bernilai 1.
   * NOT (~): Membalikkan seluruh bit (mengubah 1 menjadi 0 dan sebaliknya, menggunakan sistem 2's complement).
   * Left Shift (<<): Menggeser bit ke arah kiri sebanyak jumlah langkah tertentu (menambahkan bit 0 di kanan).
   * Right Shift (>>): Menggeser bit ke arah kanan sebanyak jumlah langkah tertentu (membuang bit di kanan).

## 2.2 Kode Program

### 2.2.1 Program "Aritmatika_2611532014"


```python
#buat file dengan nama aritmatika_nim.py
#buat program untuk operator aritmatika dalam python
#nama variabel ditambah 4 digit nim terakhir contoh : angka1_2014
#program ini menggunakan fungsi input()
#nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_2014 = int(input("Input angka-1: "))
angka2_2014 = int(input("Input angka-2: "))

# Penjumlahan
hasil_2014 = angka1_2014 + angka2_2014
print("\nOperator Penjumlahan")
print("Hasil =", hasil_2014)

# Pengurangan
hasil_2014 = angka1_2014 - angka2_2014
print("\nOperator Pengurangan")
print("Hasil =", hasil_2014)

# Perkalian
hasil_2014 = angka1_2014 * angka2_2014
print("\nOperator Perkalian")
print("Hasil =", hasil_2014)

# Pembagian, pembagian bulat, dan sisa pembagian
if angka2_2014 != 0:
    hasil_2014 = angka1_2014 / angka2_2014
    print("\nOperator Pembagian")
    print("Hasil =", hasil_2014)

    hasil_2014 = angka1_2014 // angka2_2014
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil_2014)

    hasil_2014 = angka1_2014 % angka2_2014
    print("\nOperator Sisa Pembagian")
    print("Hasil =", hasil_2014)
else:
    print("Angka kedua tidak boleh bernilai 0.")

# Pangkat
hasil_2014 = angka1_2014 ** angka2_2014
print("\nOperator Pangkat")
print("Hasil =", hasil_2014)
```

**Penjelasan :**
1. input(): Mengambil input dari pengguna.
2. int(): Mengonversi hasil masukan yang bertipe string menjadi tipe data integer agar dapat diproses dalam perhitungan matematika.
3. angka1_2014 dan angka2_2014: Variabel untuk menyimpan angka pertama dan kedua.
4. Penjumlahan (+): Menjumlahkan nilai angka1_2014 dan angka2_2014. Pengurangan (-): Mengurangi nilai angka1_2014 dengan angka2_2014. Perkalian (*): Mengalikan nilai angka1_2014 dengan angka2_2014. Hasil perhitungan disimpan ke dalam variabel hasil_2014
5. Pengondisian (if angka2_2014 != 0): Mengecek apakah nilai angka2_2014 tidak sama dengan 0. Hal ini dilakukan untuk menghindari kesalahan program (ZeroDivisionError).
6. Pembagian (/): Menghasilkan nilai pembagian riil.
7. Pembagian Bulat (//): Menghasilkan nilai pembagian yang dibulatkan ke bawah ke bilangan bulat terdekat.
8. Sisa Pembagian (%): Menghasilkan sisa dari hasil pembagian antara angka1_2014 dan angka2_2014.
9. Blok else: Akan dilakukan jika pengguna memasukkan angka 0 pada angka2_2014
10. Pangkat (**): Menghitung hasil angka1_2014 dipangkatkan dengan angka2_2014

### 2.2.2 Program "assignment_2611532014"



```python
#
#
#
#
#

angka1_2014 = int(input("Input angka-1: "))
angka2_2014 = int(input("Input angka-2: "))

print("\nNilai awal angka1_2014 =", angka1_2014)
print("Nilai angka2_2014 =", angka2_2014)

# Assignment biasa
hasil_2014 = angka1_2014
print("\nAssignment biasa (=)")
print("Hasil =", hasil_2014)

# Assignment penambahan 
hasil_2014 = angka1_2014
hasil_2014 += angka2_2014
print("\nAssignment penambahan (+=)")
print("Hasil =", hasil_2014)

# Assignment pengurangan
hasil_2014 = angka1_2014
hasil_2014 -= angka2_2014
print("\nAssignment pengurangan (-=)")
print("Hasil =", hasil_2014)

# Assignment perkalian
hasil_2014 = angka1_2014
hasil_2014 *= angka2_2014
print("\nAssignment perkalian (*=)")
print("Hasil =", hasil_2014)

# Assignment pembagian, pembagian bulat, dan sisa pembagian
if angka2_2014 != 0:
    hasil_2014 = angka1_2014
    hasil_2014 /= angka2_2014
    print("\nAssignment pembagian (/=)")
    print("Hasil =", hasil_2014)
    # Operator tambahan
    hasil_2014 = angka1_2014
    hasil_2014 //= angka2_2014
    print("\nAssignment pembagian bulat (//=)")
    print("Hasil =", hasil_2014)
    hasil_2014 = angka1_2014
    hasil_2014 %= angka2_2014
    print("\nAssignment sisa pembagian (%=)")
    print("Hasil =", hasil_2014)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assignment perpangkatan
hasil_2014 = angka1_2014
hasil_2014 **= angka2_2014
print("\nAssignment perpangkatan (**=)")
print("Hasil =", hasil_2014)
```

**Penjelasan :**
1. input() mengambil nilai dari pengguna dan int() mengonversinya dari string menjadi integer.
2. Assignment Biasa (=): Mengisi nilai hasil_2014 dengan nilai dari angka1_2014.
3. Penambahan (+=): Menjumlahkan hasil_2014 dengan angka2_2014, lalu menyimpan hasilnya kembali ke hasil_2014.
4. Pengurangan (-=): Mengurangi hasil_2014 dengan angka2_2014 dan menyimpan hasilnya kembali.
5. Perkalian (*=): Mengalikan hasil_2014 dengan angka2_2014 dan menyimpan hasilnya kembali.
6. Pengondisian (if angka2_2014 != 0): Mengecek agar tidak terjadi pembagian dengan angka nol (ZeroDivisionError).
7. Pembagian (/=): Membagi nilai hasil_2014 dengan angka2_2014 dan mengubah tipe datanya menjadi float.
8. Pembagian Bulat (//=): Membagi dan membulatkan hasilnya ke bawah (floor division).
9. Sisa Pembagian / Modulus (%=): Menyimpan sisa hasil bagi antara hasil_2014 dan angka2_2014.
10. Blok else: Dieksekusi jika pengguna memasukkan 0 pada angka2_2014 untuk menampilkan peringatan.
11. Perpangkatan (**=): Mempangkatkan nilai hasil_2014 sebesar angka2_2014

### 2.2.3 Program "bitwise_2611532014"

```python
#buat file dengan nama bitwise_nim.py
#nama variabel ditambah 4 digit terakhir contoh angka1_2014
#program ini menggunakan fungsi input()

print("\n=================================")
print("3. OPERATOR BITWISE")
print("=================================")

angka1_2014 = int(input("Masukkan angka bitwise-1: "))
angka2_2014 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalamm bentuk desimal dan biner")
print("angka1_2014 =", angka1_2014, "| biner =", bin(angka1_2014))
print("angka2_2014 =", angka2_2014, "| biner =", bin(angka2_2014))

# Bitwise AND
hasil_2014 = angka1_2014 & angka2_2014
print("\nBitwise AND (&)")
print(angka1_2014, "&", angka2_2014, "=", hasil_2014)
print("Biner hasil =", bin(hasil_2014))
print("Biner hasil (8 bit) =", format(hasil_2014, "08b"))

# Bitwise OR
hasil_2014 = angka1_2014 | angka2_2014
print("\nBitwise OR (|)")
print(angka1_2014, "|", angka2_2014, "=", hasil_2014)
print("Biner hasil =", bin(hasil_2014))
print("Biner hasil (8 bit) =", format(hasil_2014, "08b"))

# Bitwise XOR
hasil_2014 = angka1_2014 ^ angka2_2014
print("\nBitwise XOR (^)")
print(angka1_2014, "^", angka2_2014, "=", hasil_2014)
print("Biner hasil =", bin(hasil_2014))
print("Biner hasil (8 bit) =", format(hasil_2014, "08b"))

# Bitwise NOT
hasil_2014 = ~angka1_2014
print("\nBitwise NOT (~)")
print("~", angka1_2014, "=", hasil_2014)
print("Biner hasil =", bin(hasil_2014))
print("Biner hasil (8 bit) =", format(hasil_2014, "08b"))

# Bitwise geser kiri
jumlah_geser_2014 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_2014 = angka1_2014 << jumlah_geser_2014 
print("\nBitwise geser kiri (<<)")
print(angka1_2014, "<<", jumlah_geser_2014, "=", hasil_2014)
print("Biner hasil =", bin(hasil_2014))
print("Biner hasil (8 bit) =", format(hasil_2014, "08b"))

# Bitwise geser kanan
hasil_2014 = angka1_2014 >> jumlah_geser_2014 
print("\nBitwise geser kanan (>>)")
print(angka1_2014, ">>", jumlah_geser_2014, "=", hasil_2014)
print("Biner hasil =", bin(hasil_2014))
print("Biner hasil (8 bit) =", format(hasil_2014, "08b"))
```

**Penjelasan :**
1. input() dan int(): Mengambil input dari pengguna dan mengonversinya menjadi bilangan bulat desimal (integer).
2. bin(): Fungsi bawaan Python untuk menampilkan representasi biner dari suatu bilangan desimal (ditandai dengan prefiks 0b).
3. Bitwise AND (&): Membandingkan tiap bit dari dua angka. Bit hasil akan bernilai 1 hanya jika kedua bit berpasangan bernilai 1
4. Bitwise OR (|): Membandingkan tiap bit dari dua angka. Bit hasil akan bernilai 1 jika salah satu atau kedua bit bernilai 1.
5. Bitwise XOR (^): Membandingkan tiap bit dari dua angka. Bit hasil akan bernilai 1 hanya jika kedua bit bernilai beda (satu bernilai 1 dan satu lagi 0).
6. Bitwise NOT (~): membalikkan semua bit pada variabel (1 menjadi 0, dan 0 menjadi 1). Menggunakan sistem 2's complement.
7. format(hasil_2014, "08b"): Digunakan untuk menampilkan string biner sepanjang 8 bit (disertai pad angka 0 di sebelah kiri jika lebarnya kurang dari 8 bit).
8. Left Shift (<<): Menggeser seluruh bit pada angka1_2014 ke arah kiri sebanyak jumlah_geser_2014 langkah yang setara dengan mengalikan angka tersebut dengan 2 ($x \times 2^n$).
9. Right Shift (>>): Menggeser seluruh bit pada angka1_2014 ke arah kanan sebanyak jumlah_geser_2014 langkah yang setara dengan melakukan pembagian bulat dengan 2 ($x // 2^n$).

### 2.2.4 Program "lainnya_2611532014"

```python
# buat file dengan nama lainnya_nim.py
# nama variabel ditambah 4 digit nim terakhir contoh : angka1_2014
# program ini menggunakan fungsi inut()
# program operator keanggotaan dan identitas

print("=================================")
print("1. OPERATOR KEANGGOTAAN")
print("=================================")

# Input beberapa data yang diisahkan dengan koma
input_data_2014 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# mengubah input menjadi list integer
data_2014 = [int(angka.strip()) for angka in input_data_2014.split(",")]

nilai_dicari_2014 = int(input("masukkan angka yang ingin dicari: "))

# Operator in
hasil_2014 = nilai_dicari_2014 in data_2014 
print("\nOperator keanggotaan IN")
print(nilai_dicari_2014, "in", data_2014, "=", hasil_2014)

# Operator not in
hasil_2014 = nilai_dicari_2014 not in data_2014
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari_2014, "not in", data_2014, "=", hasil_2014)


print("=================================")
print("2. OPERATOR IDENTITAS")
print("=================================")

# objek1 menggunakan list dari input pengguna
objek1_2014 = data_2014

# objek2 merujuk pada objek yang sama dengan objek1
objek2_2014 = objek1_2014

# objek3 memiliki isi yang sama, tetapi merupakan objek baru
objek3_2014 = data_2014.copy()

print("objek1_2014 =", objek1_2014)
print("objek2_2014 =", objek2_2014)
print("objek3_2014 =", objek3_2014)

# Operator is
hasil_2014 = objek1_2014 is objek2_2014
print("\nOperator identitas IS")
print("objek1_2014 is objek2_2014 =", hasil_2014)

# Operator is not
hasil_2014 = objek1_2014 is not objek3_2014
print("\nOperator identitas IS NOT")
print("objek1_2014 is not objek3_2014 =", hasil_2014)

# membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1_2014 is objek3_2014 =", objek1_2014 is objek3_2014)
print("objek1_2014 == objek3_2014 =", objek1_2014 == objek3_2014)
```

**Penjelasan :**
1. split(",") memotong teks input berdasarkan tanda koma menjadi elemen-elemen terpisah.
2. [int(angka.strip()) for angka in ...] merupakan teknik list comprehension yang membersihkan spasi berlebih (strip()) dan mengonversi setiap elemen menjadi tipe data integer.
3. Hasil akhirnya disimpan dalam variabel data_2014 berbentuk kumpulan data.
4. in: Mengecek apakah nilai nilai_dicari_2014 terdapat di dalam list data_2014. Menghasilkan True jika ada, dan False jika tidak ada.
5. not in: Kebalikan dari in. Menghasilkan True jika nilai nilai_dicari_2014 tidak ada di dalam list data_2014.
6. objek1_2014 menyimpan rujukan list dari masukan pengguna.
7. objek2_2014 = objek1_2014 membuat objek2_2014 merujuk ke alokasi memori yang sama persis dengan objek1_2014.
8. objek3_2014 = data_2014.copy() membuat objek baru di lokasi memori yang berbeda, meskipun isi/nilainya sama persis dengan data_2014.
9. is: Membandingkan apakah dua variabel merujuk pada objek yang sama di memori (id(objek1_2014) == id(objek2_2014)). Bernilai True pada objek1_2014 is objek2_2014.
10. is not: Membandingkan apakah dua variabel merujuk pada objek yang berbeda di memori. Bernilai True pada objek1_2014 is not objek3_2014 karena objek3_2014 adalah hasil salinan (copy) terpisah.
11. is (Identitas): Menghasilkan False karena objek1_2014 dan objek3_2014 berada di dua alamat memori yang berbeda.
12. == (Nilai/Kandungan): Menghasilkan True karena meskipun lokasinya berbeda di memori, isi elemen dari kedua list tersebut sama persis.

### 2.2.5 Program "logika_2611532014"

```python
# buat file dengan nama logika_nim.py
#nama variabel ditambah 4 digit nim terakhir contoh : a1_2014 
#
#

# memasukkan nilai boolean
# input tidak peka terhadap huruf besar dan kecil
a1_2014 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"
a2_2014 = input("Input nilai boolean-2 (true/false): ").strip().lower() == "true"

print("\nA1 =", a1_2014)
print("A2 =", a2_2014)

# konjungsi: bernilai True jika keduanya True
hasil_2014 = a1_2014 and a2_2014
print("\nKonjungsi (AND)")
print("A1 and A2 =", hasil_2014)

# Disjungsi: bernilai True jika salah satu True
hasil_2014 = a1_2014 or a2_2014
print("\nDisjungsi (OR)")
print("A1 or A2 =", hasil_2014)

# Negasi A1: membalik nilai A1
hasil_2014 = not a1_2014
print("\nNegasi A1 (NOT)")
print("not A1 =", hasil_2014)

# Negasi A2: membalik nilai A2
hasil_2014 = not a2_2014
print("\nNegasi A2 (NOT)")
print("not A2 =", hasil_2014)

# XOR: bernilai True jika kedua nilai berbeda
hasil_2014 = a1_2014 != a2_2014
print("\nDisjungsi Eksklusif (XOR)")
print("A1 XOR A2 =", hasil_2014)
```

**Penjelasan :**
1. input(...): Menerima input teks dari pengguna.
2. .strip(): Menghapus spasi tambahan di awal maupun akhir teks input.
3. .lower(): Mengubah seluruh huruf menjadi huruf kecil agar input bersifat case-insensitive (misalnya True, TRUE, atau true diangap sama).
4. == "true": Membandingkan teks hasil olahan dengan kata "true". Jika teks yang dimasukkan adalah "true", maka perbandingan menghasilkan nilai boolean True. Jika diisi teks lain (seperti "false" atau angka), maka variabel a1_2014 dan a2_2014 akan bernilai False.
5. Operator and: Menghasilkan True hanya jika kedua operan (a1_2014 dan a2_2014) bernilai True. Jika salah satu atau keduanya False, hasilnya menjadi False.
6. Operator or: Menghasilkan True jika minimal salah satu dari operan bernilai True. Hasil hanya bernilai False jika kedua operan bernilai False.
7. Operator not: Pembalik nilai boolean (operator unary). Mengubah True menjadi False, atau sebaliknya.
8. Operasi XOR (!=): Meniru perilaku logika Exclusive OR. Menghasilkan True hanya jika nilai a1_2014 dan a2_2014 berbeda (satu True dan satu False). Jika kedua operan bernilai sama (keduanya True atau keduanya False), hasilnya adalah False.

### 2.2.6 Program "Perbandingan_2611532014"

```python
# buat file dengan nama perbandingan_nim.py
# nama variabel ditambah 4 digit nim terakhir contoh : angka1_2014
# program ini menggunakan fungsi input()
# nilai yang dimasukkan akaan dikonversi menjadi tipe data integer
# program operator perbandingan dalam python

angka1_2014 = int(input("Input angka-1: "))
angka2_2014 = int(input("Input angka-2: "))

# lebih besar dari
hasil_2014 = angka1_2014 > angka2_2014
print("\nOperator lebih besar dari")
print("angka1_2014 > angka2_2014 =", hasil_2014)

# lebih kecil dari
hasil_2014 = angka1_2014 < angka2_2014
print("\nOperator lebih kecil dari")
print("angka1_2014 < angka2_2014 =", hasil_2014)

# Lebih besar dari atau sama dengan
hasil_2014 = angka1_2014 >= angka2_2014
print("\nOperator lebih besar dari atau sama dengan")
print("angka1_2014 >= angka2_2014 =", hasil_2014)

# Lebih kecil dari atau sama dengan
hasil_2014 = angka1_2014 <= angka2_2014
print("\nOperator lebih kecil dari atau sama dengan")
print("angka1_2014 <= angka2_2014 =", hasil_2014)

# Sama dengan
hasil_2014 = angka1_2014 == angka2_2014
print("\nOperator sama dengan")
print("angka1_2014 == angka2_2014 =", hasil_2014)

# Tidak sama dengan
hasil_2014 = angka1_2014 != angka2_2014
print("\nOperator tidak sama dengan")
print("angka1_2014 != angka2_2014 =", hasil_2014)

# Tambahan: perbandingan berantai dalam Python
hasil_2014 = 0 < angka1_2014 < 100
print("\nPerbandingan berantai")
print("0 < angka1_2014 < 100 =", hasil_2014)

hasil_2014 = 0 < angka2_2014 < 100
print("0 < angka2_2014 < 100", hasil_2014)
```

**Penjelasan :**
1. input(): Menerima input teks dari pengguna melalui terminal/konsol.
2. int(): Mengonversi (casting) input teks tersebut menjadi bilangan bulat (integer) agar dapat dibandingkan secara matematis.
3. angka1_2014 dan angka2_2014: Variabel yang digunakan untuk menyimpan nilai angka pertama dan kedua.
4. (>) Lebih Besar: Menghasilkan True jika angka1_2014 bernilai lebih besar dari angka2_2014.
5. (<) Lebih Kecil: Menghasilkan True jika angka1_2014 bernilai lebih kecil dari angka2_2014.
6. (>=) Lebih Besar atau Sama Dengan: Menghasilkan True jika angka1_2014 nilainya lebih besar dari atau persis sama dengan angka2_2014.
7. (<=) Lebih Kecil atau Sama Dengan: Menghasilkan True jika angka1_2014 nilainya lebih kecil dari atau persis sama dengan angka2_2014.
8. (==) Sama Dengan: Menghasilkan True jika nilai angka1_2014 dan angka2_2014 bernilai sama persis.
9. (!=) Tidak Sama Dengan: Menghasilkan True jika nilai angka1_2014 berbeda dengan angka2_2014.
10. Perbandingan Berantai: Fitur khas Python yang memungkinkan evaluasi rentang nilai secara ringkas. Ekspresi 0 < angka1_2014 < 100 secara internal setara dengan kombinasi logika (0 < angka1_2014) and (angka1_2014 < 100).
11. Hasil Evaluasi: Menghasilkan True jika nilai angka yang diperiksa berada di dalam rentang antara 1 sampai 99 (lebih besar dari 0 dan lebih kecil dari 100).

```python
# SIMULASI TRANSAKSI DAN VALIDASI AKSES TOKO
print("=== SISTEM TRANSAKSI TOKO ===")

# INPUT DATA
nama_2014 = input("Masukkan Nama Pelanggan : ")
status_2014 = input("Masukkan Status Pelanggan (member/nonmember) : ").lower()
total_belanja_2014 = int(input("Masukkan Total Belanja : "))
jumlah_barang_2014 = int(input("Masukkan Jumlah Barang : "))
kode_promo_2014 = input("Masukkan Kode Promo : ").upper()

# DAFTAR PROMO
promo_2014 = ["HEMAT10", "HEMAT20", "MAHASISWAFTI"]

# OPERATOR PERBANDINGAN
syarat_belanja_2014 = total_belanja_2014 >= 200000
syarat_barang_2014 = jumlah_barang_2014 >= 3
status_member_2014 = status_2014 == "member"

# OPERATOR KEANGGOTAAN
promo_tersedia_2014 = kode_promo_2014 in promo_2014
promo_tidak_tersedia_2014 = kode_promo_2014 not in promo_2014

# OPERATOR LOGIKA
diskon_member_2014 = status_member_2014 and syarat_belanja_2014
promo_kelayakan_2014 = (
    syarat_belanja_2014
    and syarat_barang_2014
    and promo_tersedia_2014
)
akses_pelanggan_2014 = status_member_2014 or promo_tersedia_2014
tidak_memenuhi_promo_2014 = not promo_tersedia_2014

# OPERATOR ARITMATIKA
diskon_2014 = 0

if diskon_member_2014:
    diskon_2014 = total_belanja_2014 * 0.10
elif syarat_belanja_2014:
    diskon_2014 = total_belanja_2014 * 0.05
total_pembayaran_2014 = total_belanja_2014 - diskon_2014

harga_rata_2014 = total_belanja_2014 / jumlah_barang_2014

# OPERATOR PENUGASAN
poin_2014 = 0
poin_2014 += jumlah_barang_2014

if status_member_2014:
    poin_2014 += 10

saldo_poin_2014 = poin_2014

if promo_kelayakan_2014:
    saldo_poin_2014 -= 1

faktor_pembayaran_2014 = 1
faktor_pembayaran_2014 *= 1

nilai_per_item_2014 = total_belanja_2014
nilai_per_item_2014 /= jumlah_barang_2014

# OPERATOR IDENTITAS
objek_a_2014 = ["member"]
objek_b_2014 = ["member"]

identitas_sama_2014 = objek_a_2014 is objek_b_2014
identitas_berbeda_2014 = objek_a_2014 is not objek_b_2014
nilai_sama_2014 = objek_a_2014 == objek_b_2014

# OPERATOR BITWISE
# 0001 = Member
# 0010 = Belanja >= Rp200.000
# 0100 = Jumlah barang >= 3
# 1000 = Promo tersedia

kode_status_2014 = 0

if status_member_2014:
    kode_status_2014 |= 1

if syarat_belanja_2014:
    kode_status_2014 |= 2

if syarat_barang_2014:
    kode_status_2014 |= 4

if promo_tersedia_2014:
    kode_status_2014 |= 8

cek_member_bit_2014 = kode_status_2014 & 1
cek_promo_bit_2014 = kode_status_2014 & 8
gabungan_bit_2014 = kode_status_2014 | 8

kode_referensi_2014 = 11
perbedaan_bit_2014 = kode_status_2014 ^ kode_referensi_2014
hasil_shift_2014 = kode_status_2014 << 1

# ==========================================================
# OUTPUT
# ==========================================================

print("\n=== DATA PELANGGAN ===")
print("Nama Pelanggan   :", nama_2014)
print("Status Pelanggan :", status_2014)
print("Total Belanja    : Rp", total_belanja_2014)
print("Jumlah Barang    :", jumlah_barang_2014)
print("Kode Promo       :", kode_promo_2014)

print("\n=== HASIL VALIDASI ===")
print("Belanja >= 200000       :", syarat_belanja_2014)
print("Jumlah Barang >= 3      :", syarat_barang_2014)
print("Status Member            :", status_member_2014)
print("Kode Promo Tersedia      :", promo_tersedia_2014)
print("Mendapatkan Diskon       :", diskon_member_2014)
print("Mendapatkan Promo        :", promo_kelayakan_2014)

print("\n=== HASIL PERHITUNGAN ===")
print("Diskon           : Rp", diskon_2014)
print("Total Pembayaran : Rp", total_pembayaran_2014)
print("Rata-rata Barang  : Rp", harga_rata_2014)

print("\n=== HAK AKSES PELANGGAN ===")
print("Kode Hak Akses           :", kode_status_2014)
print("Member Access            :", status_member_2014)
print("Promo Access             :", promo_kelayakan_2014)
print("Free Shipping Access     :", promo_tersedia_2014)

print("\n=== HASIL OPERATOR ===")

print("\nOperator Perbandingan")
print("Total >= 200000          :", syarat_belanja_2014)
print("Jumlah Barang >= 3       :", syarat_barang_2014)
print("Status == member         :", status_member_2014)

print("\nOperator Logika")
print("Member AND Belanja       :", diskon_member_2014)
print("Member OR Promo          :", akses_pelanggan_2014)
print("NOT Promo                :", tidak_memenuhi_promo_2014)

print("\nOperator Penugasan")
print("Poin setelah +=          :", poin_2014)
print("Poin setelah -=          :", saldo_poin_2014)
print("Faktor setelah *=        :", faktor_pembayaran_2014)
print("Nilai per item setelah /=:", nilai_per_item_2014)

print("\nOperator Keanggotaan")
print("Kode Promo IN daftar     :", promo_tersedia_2014)
print("Kode Promo NOT IN daftar :", promo_tidak_tersedia_2014)

print("\nOperator Identitas")
print("objek_a IS objek_b       :", identitas_sama_2014)
print("objek_a IS NOT objek_b   :", identitas_berbeda_2014)
print("objek_a == objek_b       :", nilai_sama_2014)

print("\nOperator Bitwise")
print("Kode Status Biner        :", format(kode_status_2014, "04b"))
print("Kode Status Desimal      :", kode_status_2014)
print("Cek Member (& 0001)      :", cek_member_bit_2014)
print("Cek Promo (& 1000)       :", cek_promo_bit_2014)
print("Hasil OR (| 1000)        :", gabungan_bit_2014)
print("Hasil XOR (^ 1011)       :", perbedaan_bit_2014)
print("Hasil Shift (<< 1)       :", hasil_shift_2014)

print("\n=== SELESAI ===")

```

**Penjelasan**
1. Judul Program, program merupakan simulasi transaksi dan validasi akses toko. Program ini digunakan untuk mengolah data pelanggan, menghitung diskon dan total pembayaran, melakukan validasi, serta menerapkan berbagai jenis operator Python.
2. Input Data Pelanggan, program meminta beberapa data dari pengguna, yaitu nama pelanggan, status pelanggan, total belanja, jumlah barang, dan kode promo. 
3. Daftar Promo, digunakan untuk menentukan apakah kode promo yang dimasukkan oleh pelanggan tersedia atau tidak.
4. Operator Perbandingan, digunakan untuk memeriksa beberapa kondisi transaksi. Program memeriksa apakah total belanja pelanggan lebih besar atau sama dengan Rp200.000, apakah jumlah barang minimal 3, dan apakah status pelanggan sama dengan member.
5. Operator Keanggotaan, digunakan untuk memeriksa apakah kode promo yang dimasukkan terdapat dalam daftar promo yang tersedia. Operator `in` digunakan untuk memeriksa keberadaan suatu data di dalam daftar, sedangkan operator `not in` digunakan untuk memeriksa bahwa data tersebut tidak terdapat dalam daftar.
6. Operator Logika, digunakan untuk menggabungkan beberapa kondisi. Operator `and` digunakan ketika seluruh kondisi terpenuhi. Dalam program ini, pelanggan mendapatkan diskon member apabila pelanggan merupakan member dan total belanjanya memenuhi batas minimum. Operator `or` digunakan ketika salah satu kondisi sudah cukup untuk memenuhi suatu persyaratan. Operator `not` digunakan untuk membalik nilai Boolean dari suatu kondisi.
7. Operator Aritmatika, digunakan untuk melakukan perhitungan transaksi. Program menghitung jumlah diskon, total pembayaran setelah diskon, dan rata-rata harga barang. Pelanggan member yang memenuhi syarat belanja mendapatkan diskon 10%. Sementara itu, pelanggan yang bukan member tetapi memenuhi batas minimum belanja mendapatkan diskon 5%.
8. Operator Penugasan, digunakan untuk memperbarui nilai suatu variabel. Operator `+=` digunakan untuk menambahkan poin pelanggan. Operator `-=` digunakan untuk mengurangi saldo poin. Operator `*=` digunakan untuk melakukan perkalian sekaligus menyimpan hasilnya pada variabel. Operator `/=` digunakan untuk melakukan pembagian sekaligus menyimpan hasilnya pada variabel.
9. Operator Identitas, digunakan untuk membandingkan identitas dua objek. Operator `is` digunakan untuk mengetahui apakah dua variabel merujuk pada objek yang sama. Operator `is not` digunakan untuk mengetahui apakah dua variabel merujuk pada objek yang berbeda. Operator `is` memeriksa identitas objek, sedangkan `==` memeriksa kesamaan nilai.
10. Operator Bitwise, digunakan untuk mengolah data dalam bentuk bit atau bilangan biner. Program menggunakan empat bit untuk mewakili kondisi transaksi, yaitu:

    * 0001 menunjukkan pelanggan merupakan member.
    * 0010 menunjukkan total belanja minimal Rp200.000.
    * 0100 menunjukkan jumlah barang minimal 3.
    * 1000 menunjukkan kode promo tersedia.

Jika seluruh kondisi terpenuhi, maka kode status transaksi menjadi 1111 dalam bentuk biner atau 15 dalam bentuk desimal.

11. Pemeriksaan Status dengan AND. Operator bitwise AND (`&`) digunakan untuk memeriksa bagian tertentu dari kode status transaksi. Untuk memeriksa status member, kode transaksi 1111 dibandingkan dengan 0001. Hasilnya adalah 0001 atau 1 dalam desimal. Untuk memeriksa status promo, kode transaksi 1111 dibandingkan dengan 1000. Hasilnya adalah 1000 atau 8 dalam desimal. Dengan demikian, operator AND dapat digunakan untuk mengetahui apakah bit tertentu dalam kode status aktif.
12. Perbandingan Status dengan XOR. Operator XOR (`^`) digunakan untuk membandingkan dua kode biner. Kode transaksi adalah 1111, sedangkan kode referensi adalah 1011. Setelah dilakukan operasi XOR, diperoleh hasil 0100 atau 4 dalam desimal. XOR menghasilkan nilai 1 pada posisi bit yang berbeda antara kedua bilangan.
13. Operasi Shift kiri (`<<`). Operasi ini digunakan untuk menggeser bit ke sebelah kiri. Kode biner 1111 jika digeser satu posisi ke kiri menjadi 11110. Nilai tersebut dalam desimal adalah 30.

## 1.1 Latar Belakang

Perkembangan teknologi informasi sangat bergantung pada efisiensi dan keandalan sistem perangkat lunak. Di dalam pemrograman komputer, pengelolaan data merupakan proses paling penting. Setiap program komputer pada dasarnya dirancang untuk menerima data, memproses, dan menghasilkan informasi. Agar program dapat memproses data dengan benar dan efisien, komputer harus mengetahui jenis data apa yang sedang ditangani dan bagaimana data tersebut disimpan di dalam memori.

Dalam bahasa pemrograman, data disimpan menggunakan variabel dan konstanta. Variabel berfungsi sebagai tempat di dalam memori yang nilainya dapat berubah-ubah selama program berjalan, sedangkan konstanta digunakan untuk menyimpan nilai tetap yang tidak boleh diubah demi menjaga integritas data tertentu. Tipe data numerik digunakan untuk menangani perhitungan matematis dan angka, sedangkan tipe data string digunakan untuk memproses teks, karakter, dan kalimat. Selain itu, tipe data boolean digunakan untuk program yang bernilai true atau false. Input digunakan untuk menerima data atau informasi yang dimasukkan oleh pengguna ke dalam program.

## 2.1 Dasar Teori

1. Boolean
   Boolean merupakan tipe data yang digunakan untuk menyimpan nilai logika dalam pemrograman. Tipe data Boolean hanya memiliki dua nilai, yaitu `True` yang berarti benar dan `False` yang berarti salah. Boolean biasanya digunakan dalam proses pengambilan keputusan, seperti percabangan dan perulangan. Contohnya, `status = True` menunjukkan bahwa variabel `status` memiliki nilai benar.

2. Input
   Input merupakan proses untuk menerima data atau informasi dari pengguna saat program sedang dijalankan. Dalam Python, fungsi `input()` digunakan untuk meminta pengguna memasukkan suatu data. Data yang diperoleh dari `input()` secara default bertipe string. Contohnya, `nama = input("Masukkan nama: ")` digunakan untuk menerima nama yang dimasukkan oleh pengguna.

3. Konstanta
   Konstanta adalah nilai yang digunakan dalam program dan pada dasarnya tidak dimaksudkan untuk diubah selama program berjalan. Python tidak memiliki kata kunci khusus untuk membuat konstanta, sehingga biasanya konstanta ditulis menggunakan huruf kapital sebagai tanda bahwa nilainya sebaiknya tidak diubah. Contohnya, `PI = 3.14` dapat digunakan untuk menyimpan nilai tetap yang diperlukan dalam perhitungan.

4. Numerik
   Tipe data numerik digunakan untuk menyimpan data berupa angka yang dapat digunakan dalam berbagai operasi matematika. Dalam Python, tipe data numerik yang umum digunakan adalah `int` untuk bilangan bulat dan `float` untuk bilangan desimal. Contohnya, `umur = 17` menggunakan tipe `int`, sedangkan `tinggi = 165.5` menggunakan tipe `float`. Data numerik dapat digunakan untuk operasi seperti penjumlahan, pengurangan, perkalian, dan pembagian.

5. String
   String merupakan tipe data yang digunakan untuk menyimpan teks atau kumpulan karakter. String dalam Python ditulis menggunakan tanda kutip tunggal (`'...'`) atau tanda kutip ganda (`"..."`). Contohnya, `nama = "Amrina"` merupakan variabel bertipe string. String dapat digunakan untuk menyimpan berbagai informasi berupa teks, seperti nama, alamat, kalimat, dan pesan yang akan ditampilkan oleh program.
